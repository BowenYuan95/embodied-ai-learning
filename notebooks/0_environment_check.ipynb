# Lesson 0 — Environment check

Smallest possible smoke test: can this Python environment import the three
packages the project depends on, and does PyTorch see a GPU?

- `torch` — the training and tensor runtime.
- `mani_skill` — the simulator that provides `PickCube-v1`.
- `gymnasium` — the environment API that ManiSkill environments follow.

Expected output: the torch version, a CUDA boolean, and a GPU name. If CUDA is
`False`, the `get_device_name(0)` call raises instead of printing a name; that
means the environment is CPU-only and some later steps fall back to CPU.

The stored output below was recorded on 2026-09-19 when CUDA was available. It is
kept as historical evidence and is not re-executed here.


```python
import torch
import mani_skill
import gymnasium

print("Python environment OK")
print("Torch:", torch.__version__)
print("CUDA:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0))
```
